# Análise dos sinais do cenário 1
Este notebook irá carregar o arquivo `cenario_1_resultados.csv` e plotar os principais sinais do sistema, incluindo os setpoints de referência.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Carregar o arquivo CSV

file_path = 'cenario_1_resultados.csv'

df = pd.read_csv(file_path)
df.head()
```

## Gráficos dos níveis dos tanques (com referência)

```python
plt.figure(figsize=(12,6))
plt.plot(df['tempo'], df['hA'], label='hA')
plt.plot(df['tempo'], df['hB'], label='hB')
plt.plot(df['tempo'], df['hC'], label='hC')
plt.plot(df['tempo'], df['hD'], label='hD')
plt.plot(df['tempo'], df['hE'], label='hE')
plt.plot(df['tempo'], df['hC_ref'], '--', label='hC_ref')
plt.plot(df['tempo'], df['hD_ref'], '--', label='hD_ref')
plt.plot(df['tempo'], df['hE_ref'], '--', label='hE_ref')
plt.xlabel('Tempo [s]')
plt.ylabel('Nível [m]')
plt.title('Níveis dos tanques e referências ao longo do tempo')
plt.legend()
plt.grid(True)
plt.show()
```

## Gráficos das concentrações (com referência)

```python
plt.figure(figsize=(12,6))
plt.plot(df['tempo'], df['CC'], label='CC')
plt.plot(df['tempo'], df['CD'], label='CD')
plt.plot(df['tempo'], df['CE'], label='CE')
plt.plot(df['tempo'], df['CC_ref'], '--', label='CC_ref')
plt.plot(df['tempo'], df['CD_ref'], '--', label='CD_ref')
plt.plot(df['tempo'], df['CE_ref'], '--', label='CE_ref')
plt.xlabel('Tempo [s]')
plt.ylabel('Concentração [kg/m³]')
plt.title('Concentrações e referências ao longo do tempo')
plt.legend()
plt.grid(True)
plt.show()
```

## Gráficos dos comandos aplicados

```python
plt.figure(figsize=(12,6))
plt.plot(df['tempo'], df['uA'], label='uA')
plt.plot(df['tempo'], df['uB'], label='uB')
plt.plot(df['tempo'], df['uC1'], label='uC1')
plt.plot(df['tempo'], df['uC2'], label='uC2')
plt.plot(df['tempo'], df['uC3'], label='uC3')
plt.plot(df['tempo'], df['uD1'], label='uD1')
plt.plot(df['tempo'], df['uD2'], label='uD2')
plt.plot(df['tempo'], df['uD3'], label='uD3')
plt.plot(df['tempo'], df['uE1'], label='uE1')
plt.plot(df['tempo'], df['uE2'], label='uE2')
plt.plot(df['tempo'], df['uE3'], label='uE3')
plt.xlabel('Tempo [s]')
plt.ylabel('Comando')
plt.title('Comandos aplicados ao longo do tempo')
plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1))
plt.grid(True)
plt.show()
```
